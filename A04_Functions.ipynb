Write two functions:

c_to_f(celsius) converting °C to °F using the formula F = C * 9/5 + 32

f_to_c(fahrenheit) converting °F to °C using C = (F - 32) * 5/9

Test both on a few values.

```python
# TODO: Implement c_to_f and f_to_c
def c_to_f(celsius):
    return (celsius * 9/5) + 32

def f_to_c(fahrenheit):
    return (fahrenheit - 32) * 5/9


print(c_to_f(0))   # -> 32
print(f_to_c(212)) # -> 100

```

## Problem 2 — BMI with keyword-only flag
Write `bmi(weight_kg, height_m, *, rounded=False)` and return BMI value, optionally rounded to 1 decimal place when `rounded=True`.
Test it with a couple examples, including weight_kg = 70, height_kg = 1.82 (once rounded and once without rounding).

Recall the formula for bmi: weight_kg/(height_m)^2.

```python
# TODO: Define bmi(...) and test with a couple examples, including weight_kg = 70, height_kg = 1.82. Do this once with rounding and once without rounding
def bmi(weight_kg, height_m, *, rounded=False):
    result = weight_kg / (height_m ** 2)
    return round(result, 1) if rounded else result

print(bmi(70, 1.82))
print(bmi(70, 1.82, rounded=True))

```

## Problem 3 — Min/Max tuple
Write `minmax(values)` returning a `(min_value, max_value)` tuple. Apply it to a concrete input and demonstrate tuple unpacking.

```python
# TODO: Define minmax(values) and show unpacking: lo, hi = minmax([...])

#Choose a list of test values
l = [1,5,-8,3,7,-26,35,-2,0,4] 

#Define minmax(values)
def minmax(values):
    return (min(values), max(values))

#Show unpacking
lo, hi = minmax(l)

#check unpacking
print("The minimum value in", l, "is", lo)
print("The maximum value in", l, "is", hi)
```

## Problem 4 — Quadratic roots
Implement `quad_roots(a, b, c)` returning the tuple of real roots of $ax^2+bx+c=0$ (handle 0, 1, or 2 real roots).  
(You may skip complex roots or return `None` if discriminant < 0.)

Test your function on the following polynomials:
* $3x^2-12x+12$
* $x^2-2x-3$
* $x^2 -2x + 3$

```python
# TODO: Implement quad_roots(a, b, c) for real roots. Use math.sqrt for discriminant >= 0.
import math

def quad_roots(a, b, c):
    discriminant = b**2 - 4*a*c
    if discriminant < 0:
        return None  # No real roots
    root1 = (-b + math.sqrt(discriminant)) / (2*a)
    root2 = (-b - math.sqrt(discriminant)) / (2*a)
    return (root1, root2)




```

```python
print(quad_roots(3,-12,12))
print(quad_roots(1,-2,-3))
print(quad_roots(1,-2,3))
```
